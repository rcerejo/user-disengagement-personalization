## Interpretation

Random exposure helps us reason about exposure bias because it shows how engagement changes when videos are not fully selected by the production recommender. However, random exposure is not the same as randomizing users into our proposed SVD-informed personalization treatment.

Therefore, we should not estimate a causal retention effect from this dataset. The correct causal next step is the A/B test designed in Phase 8.

## Key Takeaways

- The proposed personalization treatment was not randomized in KuaiRand-Pure.
- Random exposure supports exposure-bias discussion, not retention impact claims.
- Propensity scores or regression adjustment would not solve unobserved confounding.
- Declining activity remains a predictive signal, not a proven cause of disengagement.
- Saying no to causal inference is the correct statistical decision here.
