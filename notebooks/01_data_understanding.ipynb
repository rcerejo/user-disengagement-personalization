## Interpretation

Standard recommendation logs had much higher click and long-view rates than random exposure logs. That suggests the production recommender selected more engaging videos than random exposure, but the result should be interpreted descriptively rather than as a causal retention claim.

## Key Takeaways

- The raw unit of analysis is an event-level user-video interaction.
- Standard and random logs should be kept separate because they come from different exposure mechanisms.
- User activity is skewed, so medians and percentiles matter.
- Observed non-return should be treated as disengagement, not literal app churn.
- Later modeling must preserve temporal order to avoid leakage.
