# Phase 3: Behavioral Feature Engineering

## Objective

Create a user-level modeling table that uses early-period behavior to predict future observed disengagement.

## Questions Being Answered

- What user behaviors should plausibly predict disengagement?
- Which features can be computed before the prediction window?
- Which candidate features could create leakage?
- What does the future disengagement label look like?

## Inputs

- `data/raw/KuaiRand-Pure/data/log_standard_4_08_to_4_21_pure.csv`

## Methods Used

- User-level aggregation
- Observation-window and prediction-window separation
- Recency, frequency, intensity, diversity, and activity-change features
- Leakage-aware feature selection


The modeling table has one row per user observed in the early standard recommendation period. Features are built only from early-period behavior. The label is whether the user is not observed again in the late standard recommendation period.

## Feature Families

- Frequency: number of events, active days, and events per active day.
- Recency: days since the user's last event before the observation window ended.
- Engagement quality: click rate, long-view rate, like rate, and deep-engagement rate.
- Negative feedback: hate rate.
- Diversity: number of unique videos, authors, tags, and entropy of tags/upload types.
- Repeat behavior: repeat-video and repeat-author rates.
- Change over time: second-half activity compared with first-half activity.


## Interpretation

The generated modeling table has 26,210 users and 44 columns. The future disengagement label is imbalanced: 1,200 users, or about 4.6%, are not observed in the late standard period.

That imbalance matters because Phase 4 should not rely on accuracy as the main metric. A model could predict every user returns and still appear highly accurate while failing to identify the at-risk users we care about.

## Key Takeaways

- Phase 3 creates the supervised-learning table needed for churn/disengagement modeling.
- The time split is explicit: early behavior predicts late observed disengagement.
- Aggregate video-statistics features are excluded for now because they are leakage-prone.
- The target class is rare, so Phase 4 must use imbalance-aware evaluation metrics.
- Feature interpretation must remain predictive, not causal.
