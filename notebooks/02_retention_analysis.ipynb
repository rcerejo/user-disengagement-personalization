# Phase 2: Cohort and Retention Analysis

## Objective

Understand how observed return behavior differs across user cohorts in the KuaiRand-Pure standard recommendation logs.

## Questions Being Answered

- Which users observed in the early standard period return in the later standard period?
- Does early activity level correlate with later observed return?
- How does retention vary by user active degree and first observed date?
- What censoring and selection-bias issues affect interpretation?

## Interpretation

Early activity level is strongly associated with later observed return. The lowest early-activity quartile had an observed return rate of 85.8%, while the highest early-activity quartile had an observed return rate of 99.8%.

This is descriptive evidence, not causal evidence. High activity may be a signal of stronger user interest, better recommendation fit, or other user-level differences. We should not claim that forcing more early events would cause higher retention.

## Key Takeaways

- Activity cohorts show large differences in observed return.
- First-observed-date cohorts are not true signup cohorts.
- Censoring matters because users first seen later have less future time to return.
- Survival analysis is not the main method yet because the dataset does not contain true churn events and the observed return rate is very high.
- Phase 3 should create user-level behavioral features using only information available before the prediction window.
