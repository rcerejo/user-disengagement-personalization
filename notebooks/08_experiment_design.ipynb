## Interpretation

The proposed experiment randomizes users into the current recommendation experience or a personalized strategy informed by collaborative filtering and disengagement-risk segments. The primary metric is 7-day observed return rate.

Using the observed 95.4% return-rate baseline, detecting a 1 percentage-point absolute lift with alpha 0.05 and 80% power would require about 6,140 users per group. Detecting smaller effects requires many more users.

## Key Takeaways

- Offline recommendation gains identify what is worth testing, not what has causal impact.
- The correct randomization unit is the user because retention is user-level.
- A single primary metric avoids ambiguous launch decisions.
- Guardrails protect against improving retention while worsening user experience.
- Sample size depends heavily on the minimum detectable effect.
