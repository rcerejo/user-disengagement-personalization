## Resume Bullets

- Analyzed 2.6M+ KuaiRand-Pure short-video recommendation events to study engagement, retention, and user disengagement using product metrics, cohort analysis, and leakage-aware feature engineering.

- Trained churn and recommendation models using logistic regression, random forest, gradient boosting, popularity ranking, and SVD matrix factorization; improved churn PR-AUC from a 0.046 baseline to 0.233 and SVD Hit Rate@10 from 17.6% to 22.7%.

## Interview Defense

The most important defense is evidence discipline. You can say the project produced descriptive retention findings, predictive churn modeling results, and offline recommendation improvements. You should not say it proved personalization improves retention.

Your strongest interview themes are leakage prevention, class-imbalance-aware evaluation, recommender baselines, cold start, and causal humility.

## Mock Interview Starting Point

Question: Walk me through the project end to end, and be specific about which findings are descriptive, predictive, and causal.

Strong answer: Describe the business problem, data, product metrics, retention cohorts, feature engineering, churn models, recommendation models, experiment design, and causal assessment. End by saying the final recommendation is to test segment-aware personalization, not claim it already improved retention.
