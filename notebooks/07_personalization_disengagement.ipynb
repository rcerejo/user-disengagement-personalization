## Interpretation

SVD personalization improved offline ranking more clearly for users with more early-period activity. Low-activity users had weak or negative NDCG lift, which is consistent with a cold-start problem: collaborative filtering needs enough user history to learn preferences.

This does not mean personalization caused better retention. It means the offline recommender recovered future observed interests better for some segments than others.

## Product Hypotheses

- Use collaborative filtering for users with enough behavioral history.
- Use exploration, popularity, or content-based recommendation strategies for low-history users.
- Consider segment-specific personalization tests rather than one global recommender change.
- Test retention impact through an A/B experiment before making launch claims.

## Key Takeaways

- Recommendation quality cannot be directly evaluated for fully disengaged users because they have no future positives.
- SVD lift over popularity is strongest for high-activity users.
- Low-activity users remain a cold-start challenge.
- Offline personalization gains are useful evidence, but not causal retention evidence.
- Phase 8 should design an A/B test to measure whether personalization improves retention.
