## Interpretation

The disengagement label is rare: about 4.6% of users are positive. A naive model that predicts every user returns would have high accuracy but zero recall for at-risk users.

Random forest produced the strongest ranking metrics in this first pass, with ROC-AUC around 0.867 and PR-AUC around 0.233. PR-AUC should be compared with the disengagement prevalence of about 0.046, so the model is meaningfully better than random ranking.

At a top-10% intervention threshold, random forest caught about 50.7% of disengaged users with about 23.2% precision. This is a product-style framing: if the team can only intervene on a limited group, the model can prioritize higher-risk users.

## Key Takeaways

- The models can rank users by disengagement risk better than baseline.
- Accuracy is not a useful primary metric because disengagement is rare.
- PR-AUC and top-share threshold metrics are more aligned with finding at-risk users.
- A 0.5 threshold is not automatically meaningful; thresholds should depend on the product action.
- Model results are predictive, not causal. They identify risk patterns but do not prove an intervention would work.
