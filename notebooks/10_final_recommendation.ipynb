# Phase 10: Final Product Recommendation

## Objective

Summarize the project as if presenting to a Product Manager and Data Science lead.

## Questions Being Answered

- What did we discover?
- Which findings are descriptive, predictive, or causal?
- What intervention should be tested?
- Which users should be targeted?
- What metrics determine success?
- What are the biggest limitations?

## Methods Used

- Evidence synthesis
- Product recommendation framing
- Limitation-aware decision making


## Recommendation

Test segment-aware personalization in a user-level A/B experiment. Use collaborative filtering for users with enough interaction history, and use cold-start strategies such as popularity, exploration, or content-based recommendations for low-history users.

The recommendation is to test, not immediately launch. The project produced descriptive, predictive, and offline recommendation evidence, but not causal proof of retention impact.

## Key Evidence

- Low early-activity users had an 85.8% observed return rate, compared with 99.8% for very-high-activity users.
- Random forest improved disengagement PR-AUC from a 0.046 prevalence baseline to 0.233.
- At the riskiest 10% of users, the random forest caught 50.7% of disengaged users with 23.2% precision.
- SVD matrix factorization improved Hit Rate@10 from 17.6% to 22.7% and NDCG@10 from 0.0368 to 0.0431.
- SVD lift was strongest for high-activity users and weakest for low-activity users.

## Key Takeaways

- The product should target elevated-risk users with sufficient history first.
- Low-history users need a separate cold-start strategy.
- Success should be measured by 7-day observed return rate with engagement-quality and negative-feedback guardrails.
- No causal retention claim should be made without the proposed A/B test.
