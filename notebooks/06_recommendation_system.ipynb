## Interpretation

The SVD matrix-factorization model outperformed the popularity baseline across most top-K metrics. At K=10, SVD improved Hit Rate from 17.6% to 22.7% and NDCG from 0.037 to 0.043.

The improvement is real but modest, which is realistic. Popularity is a strong baseline, especially in feeds where many users consume broadly appealing videos.

## Segment Takeaway

Personalization helped higher-activity users more clearly than low-activity users. That makes sense because collaborative filtering needs user history. Users with little history remain a cold-start challenge.

## Key Takeaways

- Matrix factorization beat popularity, so there is evidence that personalization adds offline ranking value.
- Recommendation quality should be evaluated with ranking metrics, not classification accuracy.
- NDCG@K is useful because rank position matters in feeds.
- Low-activity users are harder to personalize for because they provide less history.
- Offline recommendation gains do not prove retention improvement; that requires experiment design.
